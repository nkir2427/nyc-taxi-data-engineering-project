```python
from pathlib import Path
import pandas as pd

# 1. Define paths
base_dir = Path(r"C:\Users\User\nyc-taxi-data-engineering-project")
source_dir1 = base_dir / "data" / "processed"
source_dir2 = base_dir / "data" / "raw"
output_dir = base_dir / "data" 
output_dir.mkdir(parents=True, exist_ok=True)

# 2. Number of rows to keep for the small files
SAMPLE_SIZE = 5000

# 3. Locate an existing large file to sample from
# (Checks for parquet first, then csv)
parquet_files = list(source_dir2.glob("*.parquet"))
csv_files = list(source_dir1.glob("*.csv"))

if parquet_files:
    source_file = parquet_files[0]
    print(f"Reading sample from Parquet: {source_file.name}...")
    df_sample = pd.read_parquet(source_file).head(SAMPLE_SIZE)

elif csv_files:
    source_file = csv_files[0]
    print(f"Reading first {SAMPLE_SIZE:,} rows from CSV: {source_file.name}...")
    # nrows reads only the top rows without loading the multi-GB file into RAM
    df_sample = pd.read_csv(source_file, nrows=SAMPLE_SIZE)

else:
    raise FileNotFoundError(f"No .parquet or .csv files found in {source_dir}")

# 4. Save miniature Parquet and CSV files
small_parquet_path = output_dir / "yellow_tripdata_sample.parquet"
small_csv_path = output_dir / "yellow_tripdata_sample.csv"

# Export small Parquet
df_sample.to_parquet(small_parquet_path, index=False)
print(f"Created small Parquet: {small_parquet_path} ({small_parquet_path.stat().st_size / 1024:.2f} KB)")

# Export small CSV
df_sample.to_csv(small_csv_path, index=False)
print(f"Created small CSV:     {small_csv_path} ({small_csv_path.stat().st_size / 1024:.2f} KB)")

print(f"\nDone! Extracted {len(df_sample):,} rows.")
```
